# ☕ Smart Cafe Inventory System

A Python based command line application to manage cafe stock, calculate inventory value, and organize products. I built this project to master Python data structures, specifically **Lists and 2D Matrices**.

## 🚀 Features

* **Add New Items:** Store product names, prices, and quantities.
* **View Stock:** Neatly display all current inventory using formatted strings (f-strings).
* **Smart Search & Remove:** Delete items using a case-insensitive search mechanism.
* **Financial Analytics:** Automatically calculate the total financial value of all stock in the cafe.
* **Auto-Sorting:** Alphabetize the entire inventory with a single command.

## 🧠 What I Learned

Building this mini-project helped me solidify core programming concepts:
* **2D Lists (Matrices):** Storing multiple attributes (name, price, stock) inside a single master list.
* **List Methods:** Practical application of `.append()`, `.remove()`, and `.sort()`.
* **Control Flow:** Using `while True:` loops for continuous program execution and `if/elif/else` for menu navigation.
* **Data Validation:** Using boolean flags (`found = False`) and `.lower()` to create robust search functionality.
* **Type Casting:** Converting string inputs into `float()` and `int()` for mathematical operations.

   ```bash
   python main.py

````python
# ==========================================
# SMART CAFE POS & INVENTORY SYSTEM ☕🍩
# Built by: Maria Aqdas
# Description: A Python command-line Point of Sale project to manage cafe stock.
# ==========================================

# Yeh hamara fixed menu hai (Dictionary format mein)
# Isme cheezain (items) aur unki prices pehle se fix hain.
CAFE_MENU = {
    "1": {"name": "Black Coffee", "price": 250},
    "2": {"name": "Cold Coffee", "price": 350},
    "3": {"name": "Chocolate Cake", "price": 400},
    "4": {"name": "Green Tea", "price": 150},
    "5": {"name": "Chicken Sandwich", "price": 300}
}

# Hamari empty list jahan order/inventory save hogi
# Format will be: [Item Name, Price, Quantity]
inventory = []

while True:
    # Creating a clean, user-friendly Main Menu
    print("\n" + "="*45)
    print(" 🧋🥤🍷☕🍰🍩🍥 WELCOME TO MY CAFE 🧋🥤🍷☕🍰🍩🍥")
    print("="*45)
    print("1. View Current Inventory / Cart 👀")
    print("2. Select & Add Item from Menu ➕📢")
    print("3. Remove Item ⛳")
    print("4. Total Cafe Bill / Value 💰")
    print("5. Sort Items Alphabetically 🔠")
    print("6. Exit 👋")
    print("-" * 45)

    choice = input("Please enter your choice (1-6): ")

    # --- 1. VIEW INVENTORY ---
    if choice == "1":
        # Check if the list is empty first using 'not'
        if not inventory:
            print("\n❌ Inventory is currently empty.")
        else:
            print("\n" + "-"*10 + " CURRENT ITEMS " + "-"*10)
            # Loop through the 2D list to print each item
            for item in inventory:
                print(f"Item: {item[0]:<17} | Price: Rs.{item[1]:<4} | Stock: {item[2]}")

    # --- 2. SELECT & ADD ITEM (CASE-INSENSITIVE NAME SEARCH) ---
    elif choice == "2":
        print("\n--- 📜 CAFE MENU ---")
        # Menu ko print karwanay ka loop
        for key, value in CAFE_MENU.items():
            print(f"- {value['name']:<17} : Rs.{value['price']}")
        print("--------------------")
        
        # User se item ka naam input lena
        item_choice = input("Enter item name from the menu (e.g., Green Tea): ")
        found_in_menu = False
        
        # Menu mein check karna ke naam match hota hai ya nahi
        for key, value in CAFE_MENU.items():
            # .lower() dono side par laga diya taake chotay/baray letters ka masla khatam ho jaye
            if item_choice.lower() == value["name"].lower():
                selected_name = value["name"]
                fixed_price = value["price"]
                
                try:
                    # Sirf quantity input leni hai ab
                    quantity = int(input(f"How many '{selected_name}' do you want to add? "))
                    
                    # Naya item list mein add karna
                    new_product = [selected_name, fixed_price, quantity]
                    inventory.append(new_product) 
                    print(f"\n✅ '{selected_name}' (x{quantity}) added successfully!")
                except ValueError:
                    print("\n❌ Error: Please enter a valid number for quantity.")
                
                found_in_menu = True
                break # Agar item mil gaya toh loop ko yahin rok do
                
        if not found_in_menu:
            print("\n❌ Item not found in the menu. Please check the spelling.")

    # --- 3. REMOVE ITEM ---
    elif choice == "3":
        remove_name = input("\nEnter item name to remove: ")
        found = False

        for item in inventory:
            # Using .lower() makes the search case-insensitive here as well
            if item[0].lower() == remove_name.lower():
                inventory.remove(item)
                print(f"\n✅ '{item[0]}' has been removed successfully!")
                found = True
                break
          
        if not found:
            print(f"\n❌ Item '{remove_name}' was not found in the inventory.")     

    # --- 4. TOTAL CAFE VALUE ---
    elif choice == "4":
        total_value = 0
        
        # Calculating the grand total by multiplying price * stock for each item
        for item in inventory:
            item_value = item[1] * item[2] 
            total_value += item_value

        print(f"\n💰 Total Value / Grand Bill: Rs.{total_value}")  
    
    # --- 5. SORT INVENTORY ---
    elif choice == "5":
        if not inventory:
            print("\n❌ Inventory is empty. Nothing to sort.")
        else:
            # .sort() automatically sorts alphabetically by the first element (the name)
            inventory.sort()
            print("\n✅ Inventory has been sorted alphabetically!")

    # --- 6. EXIT ---
    elif choice == "6":
        print("\nClosing Cafe System. GOODBYE! 🙌\n")   
        break
    
    # --- INVALID INPUT HANDLING ---
    else:
        print("\n❌ Invalid choice. Please pick a number from 1 to 6.")
````
